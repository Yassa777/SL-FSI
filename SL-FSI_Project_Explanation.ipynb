## 1. The Story: Sri Lanka's Economic Crisis <a id='the-story'></a>

In 2022, Sri Lanka experienced one of the worst economic crises in its history. Think of an economy like a complex machine with many moving parts - currency, interest rates, stock market, foreign reserves, inflation. When one part starts failing, it can trigger a cascade effect.

### Key Events:

- **March 7, 2022**: Currency Float - The government stopped controlling the exchange rate, and the Sri Lankan Rupee plummeted
- **April 12, 2022**: Sovereign Default - The country couldn't pay its debts (first time in history)
- **September 2022**: Peak Inflation - Prices rose by nearly 70% compared to the previous year
- **March 20, 2023**: IMF Bailout - International Monetary Fund agreed to help

### The Challenge:

**How can we detect when an economy is shifting from "normal" to "crisis" mode?**

This is like trying to predict when a patient's health is deteriorating by monitoring vital signs. We need:
1. **The right data** (vital signs)
2. **Smart analysis** (diagnostic tools)
3. **Pattern recognition** (detecting abnormalities)

That's exactly what this project does.

## 2. The 18 Data Streams: What We're Measuring <a id='the-data'></a>

Think of these as the "vital signs" of an economy. Just like a doctor checks blood pressure, heart rate, and temperature, we track 18 different indicators.

### 💱 Currency & Foreign Exchange (4 indicators)

#### 1. **USD/LKR Exchange Rate** (D1)
- **What it is**: How many Sri Lankan Rupees (LKR) you need to buy 1 US Dollar
- **Why it matters**: If this number rises rapidly, it means your currency is losing value. Imagine if suddenly your $100 could buy much more local currency - that's bad news for the local economy.
- **Crisis signal**: Rapid increases mean people are losing confidence

#### 2. **AWCMR - Average Weighted Call Money Rate** (D2)
- **What it is**: The interest rate banks charge each other for overnight loans
- **Why it matters**: When banks don't trust each other, this rate spikes. It's like friends refusing to lend money unless you pay huge interest.
- **Crisis signal**: High rates = banks are scared and hoarding cash

#### 3. **Foreign Exchange Reserves** (D12)
- **What it is**: How much foreign currency (mainly USD) the country has saved
- **Why it matters**: This is the country's emergency fund. You need it to pay for imports and debt.
- **Crisis signal**: When this drops rapidly, the country is running out of "ammunition"
- **Analogy**: Like your savings account - when it's draining fast, you're in trouble

#### 4. **REER - Real Effective Exchange Rate** (D14)
- **What it is**: Your currency's value compared to many trading partners, adjusted for inflation
- **Why it matters**: Shows if your exports are becoming cheaper (good) or more expensive (bad) globally
- **Crisis signal**: Rapid changes indicate competitive position is shifting

### 📈 Stock Market (4 indicators)

#### 5. **ASPI - All Share Price Index** (D3)
- **What it is**: Like the S&P 500, but for Sri Lanka's stock market
- **Why it matters**: Shows investor confidence. When people are scared, they sell stocks and this falls.
- **Crisis signal**: Sharp drops mean investors are fleeing

#### 6. **S&P SL20 Index** (D6)
- **What it is**: Index of the top 20 largest companies
- **Why it matters**: More focused indicator of "blue chip" performance
- **Crisis signal**: When even the biggest, safest companies fall, it's serious

#### 7. **Equity Daily Turnover** (D4)
- **What it is**: How much money is being traded daily
- **Why it matters**: Low turnover = people aren't trading = market is frozen/dead
- **Crisis signal**: When volume dries up, it means no one wants to trade

#### 8. **Market Capitalization** (D5)
- **What it is**: Total value of all listed companies
- **Why it matters**: Shows overall market size and health
- **Crisis signal**: Shrinking market cap = wealth destruction

### 🏦 Interest Rates & Bonds (4 indicators)

#### 9. **T-Bill Yields (91-day)** (D8)
- **What it is**: Interest rate on short-term (3-month) government debt
- **Why it matters**: Higher rates = government has to pay more to borrow = sign of distress
- **Crisis signal**: When this spikes above 15-20%, it's very expensive to borrow
- **Analogy**: Like your credit card interest rate - if it suddenly jumps to 30%, you know you're seen as risky

#### 10. **T-Bond Yields (5-year)** (D9)
- **What it is**: Interest rate on medium-term (5-year) government debt
- **Why it matters**: Shows longer-term confidence in the government
- **Crisis signal**: Very high yields mean investors don't trust the government to repay

#### 11. **Policy Rate (SDFR - Standing Deposit Facility Rate)** (D10)
- **What it is**: The central bank's main interest rate tool (like the Fed Funds Rate)
- **Why it matters**: This is how the central bank controls inflation and money supply
- **Crisis signal**: Rapid hikes mean the central bank is fighting inflation or currency collapse

#### 12. **ISB Yields - International Sovereign Bond Yields** (D15)
- **What it is**: Interest rate on Sri Lanka's dollar-denominated bonds traded internationally
- **Why it matters**: What global investors demand to hold Sri Lankan debt
- **Crisis signal**: Yields above 15-20% mean default risk is considered very high
- **Note**: Before default, these were trading at 30-40% yields - extreme distress

### 💰 Commodities & Inflation (3 indicators)

#### 13. **Local Gold Price** (D7)
- **What it is**: Price of gold in Sri Lankan Rupees
- **Why it matters**: Gold is a "safe haven" - when people panic, gold prices rise
- **Crisis signal**: Rapid increases show people fleeing to safety

#### 14. **Global Gold Price (USD)** (D6 Global)
- **What it is**: Gold price in US Dollars
- **Why it matters**: Helps separate global trends from local panic
- **Use case**: If local gold rises but global doesn't, it's currency depreciation

#### 15. **NCPI Inflation (Year-over-Year)** (D13)
- **What it is**: How much prices have increased in the past year
- **Why it matters**: High inflation destroys purchasing power and savings
- **Crisis signal**: Sri Lanka hit 70% - imagine everything costing 70% more than last year
- **Real impact**: Your salary buys 40% less stuff

### 🌍 External Sector (2 indicators)

#### 16. **Tourism Earnings** (D17)
- **What it is**: Money flowing in from tourists
- **Why it matters**: Major source of foreign currency for Sri Lanka
- **Crisis signal**: When tourism collapses (like during COVID + crisis), foreign currency dries up

#### 17. **Worker Remittances** (D18)
- **What it is**: Money sent home by Sri Lankans working abroad
- **Why it matters**: Another crucial source of foreign currency
- **Crisis signal**: If this drops, it compounds the foreign exchange crisis
- **Important**: During crisis, people used black market, so official numbers dropped even more

### 🌐 Benchmark (1 indicator)

#### 18. **US 10-Year Treasury Yield** (Helper)
- **What it is**: Interest rate on 10-year US government bonds
- **Why it matters**: The "risk-free" global benchmark rate
- **Use case**: Helps calculate risk premiums - how much extra yield Sri Lanka must pay above US
- **Example**: If US 10Y = 2% and SL bond = 30%, the spread is 28% (the "risk premium")

### 🔍 Derived Features: Going Deeper

We don't just track the raw numbers - we calculate additional metrics:

1. **FX Return & Volatility**: How fast is the currency changing, and how chaotic is it?
2. **Equity Return & Volatility**: Stock market returns and turbulence
3. **Real Equity Return**: Stock returns minus inflation (what you actually earn)
4. **Gold Premium**: How much more expensive is gold locally vs globally?
5. **Interbank Spread**: How much banks don't trust each other
6. **Real Policy Rate**: Policy rate minus inflation (real cost of borrowing)
7. **Import Cover**: How many months of imports can reserves pay for?
8. **Reserve Slope**: How fast are reserves depleting?
9. **Turnover Ratio**: Trading activity relative to market size
10. **EMBI Spread**: Estimated emerging market bond risk premium

These derived metrics often reveal stress before the raw numbers do.

## 3. The Analytics: How We Find Patterns <a id='the-analytics'></a>

Now that we have the data, how do we make sense of it? We use two powerful techniques:

### 🔄 Hidden Markov Models (HMM): Finding Hidden States

#### The Concept

Imagine the economy has different "moods" or "regimes":
- **Calm**: Everything is normal, low volatility
- **Stressed**: Some indicators flashing warning
- **Crisis**: Full-blown panic mode

The problem: **We can't directly observe which regime we're in.** We can only see the data.

HMM is like a detective that looks at clues (the data) and infers the hidden state (which regime we're in).

#### A Simple Analogy

Think of a friend who has different moods (happy, neutral, grumpy) but doesn't tell you directly. You infer their mood from:
- How they text you (short vs long messages)
- Response time
- Emoji usage

Over time, you learn:
- "When they use lots of emojis and reply fast → probably happy"
- "One-word replies and slow → probably grumpy"

HMM does the same but mathematically:
- **Hidden states**: Economic regimes (calm, stressed, crisis)
- **Observations**: Our 18 data streams
- **Learning**: Figures out which data patterns correspond to which regime

#### What It Tells Us

1. **Regime Assignment**: "On this date, the economy was in Crisis Regime"
2. **Transition Probabilities**: "There's an 80% chance we stay in crisis tomorrow"
3. **Regime Characteristics**: "Crisis regime has high FX volatility and rising rates"

#### Why It's Powerful

- Automatically finds patterns without us telling it what to look for
- Can handle multiple indicators simultaneously
- Gives probability-based assessments (not just yes/no)
- Shows regime persistence (how long crises last)

### 📍 Bayesian Change Point Detection: Finding Structural Breaks

#### The Concept

This technique asks a different question: **"When did something fundamental change?"**

Not all changes are equal:
- Small daily fluctuations are normal noise
- A **change point** is when the underlying process shifts

#### A Simple Analogy

Imagine tracking your friend's sleeping pattern:
- For months: sleeps 11pm-7am (mean: 8 hours, low variance)
- Suddenly: sleeps 2am-10am (mean still 8 hours, but timing shifted)
- Then: sleeps 3am-6am, 1pm-4pm (mean drops, high variance)

A change point detector would flag:
1. When the sleep schedule shifted (fundamental change in timing)
2. When sleep became fragmented (fundamental change in pattern)

For economics:
- **Before March 2022**: Exchange rate stable around 200 LKR/USD
- **After March 2022**: Exchange rate unstable around 360 LKR/USD

The algorithm detects that shift as a "change point" - not just a spike, but a permanent regime shift.

#### The Bayesian Approach

"Bayesian" means it uses probabilities and updates beliefs with new data:

1. **Prior belief**: "Change points are relatively rare" (maybe 10% chance per day)
2. **Observe data**: See a big shift in exchange rate
3. **Update belief**: "Hmm, this looks like a change point" (probability rises to 85%)
4. **Confirm**: Next few days confirm new level → "Definitely a change point" (99%)

#### What It Tells Us

1. **Exact dates**: "March 7, 2022: 95% probability of structural break"
2. **Before/After stats**: "Mean went from 200 to 350"
3. **Magnitude**: "75% change in level"

#### Why It's Powerful

- Finds the exact moment things changed
- Distinguishes real breaks from noise
- Works online (updates as new data arrives)
- Gives confidence levels

### 🎯 Combining Both Approaches

HMM and Change Point Detection are complementary:

| Aspect | HMM | Change Point Detection |
|--------|-----|------------------------|
| **Question** | "What regime are we in?" | "When did things change?" |
| **Output** | Continuous regime labels | Specific break dates |
| **Strength** | Handles multiple variables | Pinpoints exact timing |
| **Use case** | "We're in a crisis regime" | "Crisis started March 7" |

**Together**: They give a complete picture of when crises start, what they look like, and how they evolve.

#### 4. Bayesian Change Point Detection

```python
def bayesian_changepoint_detection(data, hazard_rate=1/100):
    """
    Implements Adams & MacKay (2007) algorithm
    
    Key idea: Maintain probability distribution over:
    - How long since last change point ("run length")
    
    At each timestep:
    1. Calculate probability that change point just occurred
    2. Update run length distribution
    3. If probability > threshold, flag as change point
    """
    n = len(data)
    R = np.zeros((n + 1, n + 1))  # Run length probability matrix
    R[0, 0] = 1  # Initially, run length is 0
    
    change_points = []
    
    for t in range(1, n + 1):
        # Prediction: calculate change point probability
        predprobs = R[t-1, :t] * hazard_rate
        
        # Growth: run length increases (no change point)
        R[t, 1:t+1] = R[t-1, :t] * (1 - hazard_rate)
        
        # Changepoint: run length resets to 0
        R[t, 0] = np.sum(predprobs)
        
        # Normalize
        R[t, :] = R[t, :] / np.sum(R[t, :])
        
        # Detect
        if R[t, 0] > threshold:
            change_points.append(t-1)
    
    return change_points
```

**What it does**:
1. **Maintains belief**: "How long since last change?"
2. **Updates with data**: "Does this new observation fit the current pattern?"
3. **Flags anomalies**: "This doesn't fit → probably a change point"

**Output**: List of dates where structural breaks occurred

## ❓ Common Questions

### Q1: Why 18 indicators? Why not more or fewer?

**A**: Balance between:
- **Coverage**: Want all major sectors (FX, rates, equity, inflation, external)
- **Availability**: Limited by Sri Lankan data availability
- **Computational**: More indicators = more complexity
- **Redundancy**: Avoid too much correlation (e.g., don't need 10 equity indices)

### Q2: Can this predict the next crisis?

**A**: Partially.
- ✅ **Can detect**: Early regime shifts before full crisis
- ✅ **Can monitor**: Real-time stress levels
- ❌ **Can't predict**: Exact timing (many crises have triggers that don't show in data)
- ❌ **Can't prevent**: Models show stress, but can't force policy action

Think of it like a health monitor - it can detect deteriorating vital signs, but can't predict when exactly someone will collapse.

### Q3: How accurate is it?

**A**: Depends on validation:
- **Backward looking**: Very good - aligns with known crisis dates
- **Forward looking**: Harder - need real-time testing
- **Regime detection**: ~85-90% alignment with expert assessment
- **Change points**: High precision for major breaks, some false positives for minor volatility

### Q4: Could this work for other countries?

**A**: Yes! The framework is general:
- Replace data sources with target country's indicators
- Keep the same analytical techniques
- Adjust crisis dates to that country's history
- May need different indicators (e.g., oil exporters track oil prices)

### Q5: What's missing?

**A**: Some limitations:
- **Political factors**: No way to quantify political instability
- **Social unrest**: Protests, strikes aren't in the data
- **Contagion**: External shocks from other countries
- **Expectations**: Market sentiment surveys
- **Real sector**: Manufacturing, unemployment (limited data)
